# **Qué es el análisis de sentimiento**

Es una tarea de clasificación de texto. Dado un texto, el modelo decide si la intención utilizada es positiva, negativa o neutral.

No es "**entender**" el texto como lo haría una persona. Es detectar patrones de palabras y contexto que, estadísticamente, se asocian a cada categoría.

Para esta práctica en español, el modelo más adecuado es **RoBERTuito**. Es un modelo entrenado específicamente con **500 millones de tweets en español**, lo que le permite entender jerga, emojis, abreviaturas y texto informal.

Está diseñado para redes sociales, que es donde el análisis de sentimiento tiene más aplicación real. Y funciona bien en tareas de sentimiento, odio e ironía en español.

RoBERTuito no se entrenó desde cero. Es un modelo preentrenado: alguien invirtió muchísimo tiempo y recursos en entrenarlo con textos masivos para que "entienda" el español.

Luego ese modelo base se ajustó (mediante fine-tuning) con un conjunto más pequeño de tweets etiquetados como positivos, negativos o neutros. El resultado es un modelo especializado en análisis de sentimiento en español, listo para usar.

Esto es importante porque explica por qué tú no entrenas nada. Usas un modelo que ya existe, ya está entrenado, y solo lo llamas desde Colab. Es exactamente lo que haría un desarrollador profesional: **no reinventar la rueda, usar lo que ya funciona**.

# **Hugging Face**

Hugging Face es la plataforma donde viven estos modelos. Su función **pipeline** es una abstracción: tú le dices "quiero realizar un análisis de sentimiento" y "usa este modelo", y ella se encarga de todo lo demás (preprocesar el texto, pasarlo por el modelo, devolver el resultado).

Solo se necesita entender que pipeline es el "botón" que conecta el texto con un modelo ya entrenado.

El modelo devuelve dos cosas:

    Una etiqueta: POS, NEG o NEU

    Una confianza (score), un número entre 0 y 1

La confianza no es lo mismo que **"acierto"**. Es la seguridad del modelo en su propia predicción. Un score de 0.99 no garantiza que tenga razón; un score de 0.51 en una frase ambigua es normal y no es un fallo.

El modelo no lee intenciones, lee patrones. El guiño o el signo de interrogación alteran el contexto que el modelo aprendió durante el preentrenamiento, y eso mueve la predicción.

```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis", model="pysentimiento/robertuito-sentiment-analysis")
# carga un modelo de inteligencia artificial
# entrenado para analizar el sentimiento
# de textos en español, con especial enfoque
# en lenguaje informal y redes sociales

# === este es el texto  que se clasifica ===
texto = "Me encanta este módulo de IA"
# =====================================

resultado = clasificador(texto)
print(resultado)
```

```python
resultado = clasificador("Esto es horrible.")
print(resultado)
```

**¿Cómo funciona el análisis de sentimientos? ¿Qué límites tiene?**

El análisis de sentimientos utiliza técnicas de procesamiento del lenguaje natural (PLN) para clasificar un texto según la emoción o valoración que expresa. En esta práctica, el modelo RoBERTuito analiza el texto y devuelve una etiqueta (POS, NEG o NEU) y una puntuación (*score*) que representa la confianza del modelo en esa predicción.

El modelo funciona detectando patrones aprendidos durante su entrenamiento con muchos textos en español. Tiene en cuenta las palabras y parte del contexto en el que aparecen. No comprende el texto como una persona, sino que calcula qué categoría es más probable.

Sus principales límites son:

- Puede confundir la ironía, el sarcasmo y las bromas.
- Puede interpretar mal expresiones ambiguas, dobles sentidos o frases que necesitan contexto.
- Los emojis, las faltas de ortografía, las abreviaturas y la jerga pueden cambiar el resultado.
- Puede cometer errores con expresiones regionales o temas que no aparecían suficientemente en sus datos de entrenamiento.
- La puntuación (*score*) indica confianza, no garantiza que la clasificación sea correcta.

Por tanto, sus resultados sirven como ayuda automática, pero en situaciones importantes conviene revisarlos por una persona.

# **Práctica**

Implementa en python un analizador de comentarios. Los comentarios positivos y neutros serán aceptados, salvo que contengan alguna palabra de la lista negra de palabras. Los comentarios negativos que contengan alguna palabra de la lista negra o que tengan un score de más de 0.5 serán rechazadas.

La lista negra no aparece especificada en el enunciado, así que en el código se incluye una lista de ejemplo que debes sustituir por las palabras que indique el profesor, si las ha facilitado.

```python
# Analizador de comentarios con lista negra

# IMPORTANTE: esta lista es un ejemplo. Puedes modificarla y añadir
# las palabras que indique el profesor o que quieras bloquear.
lista_negra = ["insulto1", "insulto2", "spam"]

def contiene_palabra_lista_negra(texto, palabras_prohibidas):
    """Comprueba si el texto contiene alguna palabra de la lista negra."""
    palabras_texto = texto.casefold()
    return any(palabra.casefold() in palabras_texto
               for palabra in palabras_prohibidas if palabra.strip())

def analizar_comentario(texto):
    """
    Acepta los comentarios POS y NEU salvo que contengan una palabra
    de la lista negra.

    Rechaza los comentarios NEG si contienen una palabra de la lista
    negra o si su score es estrictamente superior a 0.5.
    """
    if not isinstance(texto, str) or not texto.strip():
        return {
            "comentario": texto,
            "sentimiento": "SIN TEXTO",
            "score": None,
            "decision": "RECHAZADO",
            "motivo": "El comentario está vacío o no es válido."
        }

    # Primero se comprueba la lista negra.
    palabra_prohibida = contiene_palabra_lista_negra(texto, lista_negra)

    # Se obtiene la predicción del modelo.
    prediccion = clasificador(texto)[0]
    etiqueta = prediccion["label"].upper()
    score = float(prediccion["score"])

    if etiqueta in ("POS", "NEU"):
        if palabra_prohibida:
            decision = "RECHAZADO"
            motivo = "Contiene una palabra de la lista negra."
        else:
            decision = "ACEPTADO"
            motivo = "Comentario positivo o neutro sin palabras prohibidas."

    elif etiqueta == "NEG":
        if palabra_prohibida:
            decision = "RECHAZADO"
            motivo = "Es negativo y contiene una palabra de la lista negra."
        elif score > 0.5:
            decision = "RECHAZADO"
            motivo = "Es negativo y su score es superior a 0.5."
        else:
            decision = "ACEPTADO"
            motivo = "Es negativo, pero su score no supera 0.5."

    else:
        # Por seguridad, una etiqueta inesperada no se acepta automáticamente.
        decision = "RECHAZADO"
        motivo = f"Etiqueta no contemplada: {etiqueta}."

    return {
        "comentario": texto,
        "sentimiento": etiqueta,
        "score": round(score, 4),
        "decision": decision,
        "motivo": motivo
    }


# Probamos el analizador con varios comentarios.
comentarios = [
    "Me encanta este producto, es fantástico.",
    "El paquete llegó ayer por la tarde.",
    "No me ha gustado nada, es horrible.",
    "¡Genial! Se ha vuelto a romper...",
]

for comentario in comentarios:
    resultado = analizar_comentario(comentario)
    print(resultado)
    print("-" * 60)
```
